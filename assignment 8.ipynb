```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

text = """Artificial intelligence is changing the world.
AI helps machines learn from data and solve problems.
It is used in healthcare, education, and business.
AI makes our work faster and easier.
I enjoy learning AI and building intelligent applications."""
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("Cleaned Text:")
print(clean_text)
```

```python
words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("Word Tokens:", words)
print("\nSentence Tokens:", sentences)
```

```python
print("Using split():")
print(clean_text.split())

print("\nUsing word_tokenize():")
print(word_tokenize(text.lower()))
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in words if word not in stop_words]

print("Words after removing stopwords:")
print(filtered_words)
```

```python
frequency = FreqDist(filtered_words)

print("Word Frequency Distribution:")
print(frequency.most_common())

frequency.plot()
```

```python
texts = [
    "Artificial intelligence helps machines learn from data.",
    "AI is useful in healthcare and education.",
    "Machine learning is an important part of artificial intelligence."
]
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer()
bow = cv.fit_transform(texts)

print("Feature Names:")
print(cv.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(texts)

features = tfidf.get_feature_names_out()

print("TF-IDF Matrix:")
print(matrix.toarray())

for i in range(len(texts)):
    scores = matrix[i].toarray()[0]
    top_indices = scores.argsort()[-3:][::-1]

    print("\nText", i + 1, "Top 3 Keywords:")

    for index in top_indices:
        print(features[index], ":", round(scores[index], 3))
```

```python
text1 = """Artificial intelligence helps machines learn.
AI analyzes data and makes predictions.
It is useful in healthcare and education.
AI automates tasks and solves problems.
It improves productivity."""

text2 = """Blockchain stores information in secure blocks.
It provides transparency and data security.
Blockchain is useful in finance and banking.
It records transactions without central control.
It improves trust and integrity."""
```

```python
import re

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    return text.split()

words1 = preprocess(text1)
words2 = preprocess(text2)

print("Text 1 Tokens:", words1)
print("Text 2 Tokens:", words2)
```

```python
set1 = set(words1)
set2 = set(words2)

jaccard = len(set1.intersection(set2)) / len(set1.union(set2))

print("Jaccard Similarity:", round(jaccard, 3))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(vectors)[0][1]

print("Cosine Similarity:", round(cosine, 3))
```

```python
reviews = [
    "This product is amazing and very useful.",
    "I love the excellent quality.",
    "The product works perfectly.",
    "The product is terrible and disappointing.",
    "The quality is poor.",
    "The product is okay."
]
```

```python
%pip install textblob
```

```python
from textblob import TextBlob

positive_reviews = []

for review in reviews:
    result = TextBlob(review).sentiment

    polarity = result.polarity
    subjectivity = result.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Polarity:", round(polarity, 2))
    print("Subjectivity:", round(subjectivity, 2))
    print("Sentiment:", sentiment)
```

```python
%pip install wordcloud
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

positive_text = " ".join(positive_reviews)

wc = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.imshow(wc, interpolation="bilinear")
plt.axis("off")
plt.show()
```
