```python
#q1
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)

# a) Add 2 to all elements
print("After adding 2:", arr + 2)

# b) Multiply 3 with all elements
print("After multiplying by 3:", arr * 3)

# c) Divide every element by 2
print("After dividing by 2:", arr / 2)
```

```python
#q2b
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

value = 1

print("Most frequent value:", value)
print("Indices:", np.where(x == value)[0])
```

```python
#q2b
y = np.array([1, 1, 1, 2, 3, 4, 2, 3, 3, 3, 1])

value = 1
print("Most frequent value:", value)
print("Indices:", np.where(y == value)[0])
```

```python
#q3
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

# a) 1st row, 2nd column
print(arr[0, 1])

# b) 3rd row, 1st column
print(arr[2, 0])
```

```python
#q4
import numpy as np

pankhil = np.linspace(10, 100, 25)

print("Array:", pankhil)
print("Shape:", pankhil.shape)
print("Total elements:", pankhil.size)
print("Data type:", pankhil.dtype)
print("Total bytes:", pankhil.nbytes)

# Convert into 5 x 5
pankhil = pankhil.reshape(5, 5)

print("5 x 5 array:")
print(pankhil)

# Transpose
print("Transpose:")
print(pankhil.T)
```

```python
#q5
import numpy as np

ucs420_pankhil = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Array:")
print(ucs420_pankhil)

print("Mean:", np.mean(ucs420_pankhil))

print("Median:", np.median(ucs420_pankhil))

print("Maximum:", np.max(ucs420_pankhil))

print("Minimum:", np.min(ucs420_pankhil))

print("Unique elements:", np.unique(ucs420_pankhil))

reshaped_ucs420_pankhil = ucs420_pankhil.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_pankhil)

resized_ucs420_pankhil = np.resize(ucs420_pankhil, (2, 3))

print("Resized array:")
print(resized_ucs420_pankhil)
```
