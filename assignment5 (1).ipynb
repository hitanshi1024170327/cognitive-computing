```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

value = np.bincount(x).argmax()

print("Most frequent value:", value)
print("Indexes:", np.where(x == value)[0])
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

value = np.bincount(y).argmax()

print("Most frequent value:", value)
print("Indexes:", np.where(y == value)[0])
```

```python
import numpy as np

arr = np.linspace(1, 100, 25)

print("Array:", arr)
print("Shape:", arr.shape)
print("Total elements:", arr.size)
print("Data type:", arr.dtype)
print("Total bytes:", arr.nbytes)

transpose = arr.reshape(5, 5)

print("Transpose:", transpose.T)
```

```python
import numpy as np

ucs420_hitanshi = np.array([[10, 20, 30, 40],
                            [50, 60, 70, 80],
                            [90, 15, 20, 35]])

print("Mean:", np.mean(ucs420_hitanshi))
print("Median:", np.median(ucs420_hitanshi))
print("Maximum:", np.max(ucs420_hitanshi))
print("Minimum:", np.min(ucs420_hitanshi))

reshaped_ucs420_hitanshi = ucs420_hitanshi.reshape(4, 3)

print("Reshaped array:")
print(reshaped_ucs420_hitanshi)

resized_ucs420_hitanshi = ucs420_hitanshi.reshape(2, 6)

print("Resized array:")
print(resized_ucs420_hitanshi)
```
