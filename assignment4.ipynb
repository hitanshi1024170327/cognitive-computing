```python
import pandas as pd

# Take roll number
roll_no = input("Enter your roll number: ")

# Get last two digits
d1 = int(roll_no[-2])
d2 = int(roll_no[-1])

# Categories
categories = ["billing", "account", "general"]

# Find category for each digit
cat1 = categories[d1 % 3]
cat2 = categories[d2 % 3]

# Fixed entries
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Personalized entry 1
entry1 = {
    "question": "how do i update my registered mobile number",
    "answer": "You can update your mobile number from account settings.",
    "keywords": "mobile number update",
    "category": cat1
}

# Personalized entry 2
entry2 = {
    "question": "how can i check my account details",
    "answer": "You can check your account details in the profile section.",
    "keywords": "account details profile",
    "category": cat2
}

# Add personalized entries
fixed_entries.append(entry1)
fixed_entries.append(entry2)

# Create DataFrame
df = pd.DataFrame(fixed_entries)

# Print final DataFrame
print("\nFinal FAQ DataFrame:")
print(df)
```

```python
def score_query(query, df):

    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():

        text = row["question"] + " " + row["keywords"]

        text_words = text.lower().split()

        score = 0

        for word in query_words:
            if word in text_words:
                score += 1

        if score > 0:
            results.append((score, row["question"]))

    # Sort by score from highest to lowest
    results.sort(reverse=True)

    return results


# Test the function
query = input("Enter your question: ")

results = score_query(query, df)

print("\nMatching entries:")

for score, question in results:
    print("Score:", score, "| Question:", question)
```

```python
def same_category(category_name, df):

    result = df[df["category"] == category_name]

    return result


# Take category of first personalized entry
category_name = entry1["category"]

print("\nCategory:", category_name)

print("\nQuestions belonging to this category:")

result = same_category(category_name, df)

print(result[["question", "category"]])
```

```python
# Display all entries
print(df)

# Ask which entry to modify
choice = int(input("\nEnter row number to update (0-5): "))

# Ask for new keyword
new_keyword = input("Enter new keyword: ")

# Add keyword
df.loc[choice, "keywords"] = df.loc[choice, "keywords"] + " " + new_keyword

# Display updated DataFrame
print("\nUpdated DataFrame:")
print(df)

# Save to CSV
filename = roll_no + "_faq_data.csv"

df.to_csv(filename, index=False)

print("\nFile saved as:", filename)
```

```python
category_count = df.groupby("category").size()

print("\nNumber of FAQ entries per category:")
print(category_count)
```

```python
def best_matches(query, df):

    query_words = query.lower().split()

    scores = []

    for index, row in df.iterrows():

        text = row["question"] + " " + row["keywords"]

        text_words = text.lower().split()

        score = 0

        for word in query_words:
            if word in text_words:
                score += 1

        scores.append((score, row["question"]))

    # Find highest score
    highest_score = max(score for score, question in scores)

    print("\nHighest score:", highest_score)

    print("\nBest matching entries:")

    for score, question in scores:

        if score == highest_score:
            print("Score:", score, "| Question:", question)


# -------------------------
# Tie example
# -------------------------

print("\n--- TIE EXAMPLE ---")

best_matches("fee", df)


# -------------------------
# Non-tie example
# -------------------------

print("\n--- NON-TIE EXAMPLE ---")

best_matches("password", df)
```
