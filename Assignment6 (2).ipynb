```python
#Q1
import numpy as np
import matplotlib.pyplot as plt

M = float(input("Enter M: "))

x = np.linspace(-10, 10, 100)

y1 = M * x**2
y2 = M * np.sin(x)

plt.plot(x, y1, 'r-', label='y = M*x^2')
plt.plot(x, y2, 'b--', label='y = M*sin(x)')

plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

```python
#Q2
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create dataset
data = {
    'Subject': ['Maths', 'Physics', 'Chemistry', 'English', 'Computer'],
    'Score': [85, 78, 90, 88, 95]
}

# Convert into DataFrame
df = pd.DataFrame(data)

print(df)

# Create bar plot
sns.barplot(x='Subject', y='Score', data=df, hue='Subject', palette='Set2', legend=False)

# Add scores on top of each bar
for i in range(5):
    plt.text(i, df['Score'][i] + 1, df['Score'][i], ha='center')

# Title and labels
plt.title("Scores of Five Subjects")
plt.xlabel("Subjects")
plt.ylabel("Scores")

# Add grid
plt.grid(axis='y')

plt.show()
```

```python
#Q3
import numpy as np
import matplotlib.pyplot as plt

# Enter roll number
roll_no = int(input("Enter your roll number: "))

# Set roll number as seed
np.random.seed(roll_no)

# Generate 50 random values
data = np.random.randn(50)

# Calculate cumulative sum
cumulative_sum = np.cumsum(data)

# Create 2x2 subplot
plt.figure(figsize=(10, 7))

# 1. Line plot - Cumulative Sum
plt.subplot(2, 2, 1)
plt.plot(cumulative_sum)
plt.title("Cumulative Sum")
plt.xlabel("Index")
plt.ylabel("Sum")
plt.grid(True)

# 2. Scatter plot - Random Noise
plt.subplot(2, 2, 2)
plt.scatter(range(50), data)
plt.title("Random Noise")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)

# Show the plots
plt.tight_layout()
plt.show()
```
